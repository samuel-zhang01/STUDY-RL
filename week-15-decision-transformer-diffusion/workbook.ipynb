## Statistical investigation — Conditioning on a return is not intervening on an outcome

**Goal.** Separate achieved-return conditioning from a causally effective action. This bounded reference investigation runs on CPU without downloads.
Write your prediction in a new Markdown cell before executing it.

**Predict.** Half the trajectories come from easy tasks and half from hard tasks.
Success probabilities are 0.8 and 0.2, respectively; a recorded action has no
effect in this fixture. Predict P(easy | success). Would conditioning a
sequence model on successful trajectories identify an action that causes success?

**Work a small case by hand.** Let E denote an easy task and S success. The marginal success rate is
P(S)=0.5×0.8+0.5×0.2=0.5. Bayes' rule gives
P(E|S)=P(S|E)P(E)/P(S)=0.8×0.5/0.5=0.8. The random action is independent
of task difficulty and has no effect here, so both interventions do(A=0)
and do(A=1) still have success probability 0.5.

**Reading bridge:** [Probability and statistics lesson](../docs/statistics/01-probability-conditioning.md).
These are original course examples informed by the textbook sections in the
[statistics source catalog](../docs/statistics/SOURCES.md); textbook PDFs are not required to run them.

**Requested versus achieved return.** The simulation conditions on achieved
success, not a desired-return input token. A desired-return token would enter
a separate policy mechanism, which this fixture does not implement. Conditioning
on achieved success makes its success probability one by definition; the
nontrivial calculation here concerns the selected easy-task composition.


```python
import numpy as np
import matplotlib.pyplot as plt
from common.plotting import RL_PALETTE, use_rl_style
use_rl_style()
rng_stats=np.random.default_rng(115);n_trajectories=4000
easy=rng_stats.binomial(1,.5,n_trajectories).astype(bool)
action=rng_stats.binomial(1,.5,n_trajectories)
success=rng_stats.binomial(1,np.where(easy,.8,.2)).astype(bool)
print('P(easy):',easy.mean(),'P(easy | success):',easy[success].mean(),'Bayes prediction: .8')
print('Observed success by random action:',[success[action==a].mean() for a in [0,1]])
fig,axes=plt.subplots(1,2,figsize=(9,3.3))
axes[0].bar(['All trajectories','Successful only'],[easy.mean(),easy[success].mean()])
axes[0].set(ylabel='Easy-task fraction',ylim=(0,1))
axes[1].bar(['Action 0','Action 1'],[success[action==a].mean() for a in [0,1]])
axes[1].set(ylabel='Observed success rate',ylim=(0,1))
fig.suptitle('Return conditioning changes the composition of the data')
fig.tight_layout();plt.show()
```

**Interpret and transfer.** Successful trajectories contain more easy tasks. This fixture's actions
are randomized and causally irrelevant, so changing action cannot reproduce
the conditional success distribution by itself. A useful return-conditioned
policy still needs support and an environment where its actions affect outcomes.

**Transfer.** Make action 1 increase success within each difficulty stratum.
Specify the new mechanism, then distinguish conditioning on an achieved
success return, a requested-return token as policy input, observed
P(success | action), and P(success | do(action)) under the simulator.


**Read your result.** The successful-only bar should be near 0.8 easy tasks, while the two action
success bars remain near 0.5. Conditioning selected a different composition;
it did not discover a causally effective action.
