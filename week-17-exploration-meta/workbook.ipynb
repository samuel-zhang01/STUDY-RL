## Statistical investigation — Choosing the best noisy run creates selection optimism

**Goal.** Measure the optimism introduced by selecting a noisy validation maximum. This bounded reference investigation runs on CPU without downloads.
Write your prediction in a new Markdown cell before executing it.

**Predict.** Twenty hyperparameter settings have identical true value zero. Each has a
noisy validation estimate. Select the largest one, then evaluate that chosen
setting with fresh noise. Predict the average validation–test gap over 500
independent repetitions of this entire selection experiment.

**Work a small case by hand.** For just two equally good settings, validation estimates Z₁,Z₂ are
independent N(0,0.3²). The identity for a noisy maximum gives
E[max(Z₁,Z₂)]=0.3/sqrt(pi)≈0.1693, although both true values are zero.
If J is selected using validation, independent test noise satisfies
E[test_J|J]=0, hence E[test_J]=0. Searching 20 settings amplifies the
selection effect without changing any true setting value.

**Reading bridge:** [Probability and statistics lesson](../docs/statistics/08-adaptive-data.md).
These are original course examples informed by the textbook sections in the
[statistics source catalog](../docs/statistics/SOURCES.md); textbook PDFs are not required to run them.


```python
import numpy as np
import matplotlib.pyplot as plt
from common.plotting import RL_PALETTE, use_rl_style
use_rl_style()
rng_stats=np.random.default_rng(117)
validation=rng_stats.normal(0,.3,(500,20));test=rng_stats.normal(0,.3,(500,20))
chosen=validation.argmax(axis=1)
selected_validation=validation[np.arange(500),chosen]
selected_test=test[np.arange(500),chosen]
print('Selected validation mean:',selected_validation.mean(),'fresh evaluation mean:',selected_test.mean())
fig,axes=plt.subplots(1,2,figsize=(9,3.3))
shared_bins=np.linspace(min(selected_validation.min(),selected_test.min()),max(selected_validation.max(),selected_test.max()),28)
axes[0].hist(selected_validation,bins=shared_bins,alpha=.5,label='Chosen validation estimate')
axes[0].hist(selected_test,bins=shared_bins,alpha=.5,label='Fresh selected-setting evaluation')
axes[0].set(xlabel='Estimated performance',ylabel='Independent selection experiments');axes[0].legend(fontsize=8)
ks=[1,2,5,10,20,50]
optimism=[rng_stats.normal(0,.3,(2000,k)).max(axis=1).mean() for k in ks]
axes[1].plot(ks,optimism,'o-');axes[1].set(xlabel='Number of settings searched',ylabel='Selected validation mean\n(true value 0)')
fig.tight_layout();plt.show()
```

**Interpret and transfer.** Selecting a noisy maximum couples the reported estimate to its noise.
Fresh test data assess the selection rule after it is frozen. Repeatedly
editing the rule based on the test result turns that test into validation.

**Transfer.** Replace settings by meta-learning tasks. State whether you want
uncertainty over adaptation episodes, training seeds, or new tasks, and design
the outer evaluation split accordingly.


**Read your result.** The selected-validation histogram shifts right while fresh evaluations
remain centered near zero. The right panel varies search size, so its
optimism is a property of the selection procedure, not a new true performance gain.
