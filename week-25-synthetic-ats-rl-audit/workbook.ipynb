## Statistical investigation — Bootstrap the entire logged session

**Goal.** Preserve session-level numerators and denominators in a SNIPS bootstrap. This bounded reference investigation runs on CPU without downloads.
Write your prediction in a new Markdown cell before executing it.

**Predict.** A fixed target is evaluated from 60 independent sessions with five related
logged decisions each. Session-level reward propensity creates dependence.
Compare a row bootstrap with a whole-session bootstrap of the SNIPS ratio.
When resampling a cluster, retain its numerator and denominator together.

**Work a small case by hand.** The observed-action weights are 0.8/0.5=1.6 and 0.2/0.5=0.4.
For a small five-row-per-session arithmetic anchor, let session 1 have
weighted numerator/denominator (1.6,2) and session 2 have (1.6,8).
The pooled SNIPS ratio is (1.6+1.6)/(2+8)=0.32. Averaging the two
session ratios gives (0.8+0.2)/2=0.5, a different estimand.
Resampling whole sessions must retain each numerator and denominator together.

**Reading bridge:** [Probability and statistics lesson](../docs/statistics/07-importance-weighting.md).
These are original course examples informed by the textbook sections in the
[statistics source catalog](../docs/statistics/SOURCES.md); textbook PDFs are not required to run them.


```python
import numpy as np
import matplotlib.pyplot as plt
from common.plotting import RL_PALETTE, use_rl_style
use_rl_style()
rng_stats=np.random.default_rng(125);sessions=60;rows_per_session=5
latent=rng_stats.beta(2,2,sessions);actions=rng_stats.binomial(1,.5,(sessions,rows_per_session))
reward=rng_stats.binomial(1,np.broadcast_to(latent[:,None],actions.shape))
target_prob=np.where(actions==1,.8,.2);weights=target_prob/.5
num=weights*reward;den=weights
point=num.sum()/den.sum();row_draws=[];session_draws=[]
for _ in range(600):
    row_idx=rng_stats.integers(0,num.size,num.size)
    row_draws.append(num.ravel()[row_idx].sum()/den.ravel()[row_idx].sum())
    session_idx=rng_stats.integers(0,sessions,sessions)
    session_draws.append(num[session_idx].sum()/den[session_idx].sum())
print('Fixed-target SNIPS:',point,'independent sessions:',sessions)
print('Row-IID 95% percentile interval (wrong unit):',np.quantile(row_draws,[.025,.975]))
print('Whole-session 95% percentile interval:',np.quantile(session_draws,[.025,.975]))
fig,ax=plt.subplots(figsize=(7,3.5))
shared_bins=np.linspace(min(min(row_draws),min(session_draws)),max(max(row_draws),max(session_draws)),32)
ax.hist(row_draws,bins=shared_bins,alpha=.5,label='Rows (invalid IID assumption)')
ax.hist(session_draws,bins=shared_bins,alpha=.5,label='Intact independent sessions')
ax.axvline(point,color=RL_PALETTE['gray'],ls='--');ax.set(xlabel='Bootstrapped SNIPS ratio',ylabel='Bootstrap draws')
ax.legend(fontsize=8);fig.tight_layout();plt.show()
```

**Interpret and transfer.** A bootstrap repeats the empirical sampling design; it cannot repair an
incorrect logger probability, unobserved confounding or absent support. This
fixture's nonzero target probabilities avoid undefined bootstrap denominators.
Rare-overlap designs require reporting failed resamples instead of silently
replacing them with zero.

**Transfer.** Let a target depend on a reward model fitted on the same log.
Separate policy/nuisance fitting from evaluation before choosing a resampling
scheme. Explain the stronger complication if logging itself adapts over time.


**Read your result.** Compare the two bootstrap distributions and their printed interval widths.
Both reuse the same fixed target and recorded propensities; the whole-session
procedure preserves the source of dependence encoded by the latent session rate.

<details><summary>Worked feedback for the changed assumption</summary>

A target or nuisance model fitted on the evaluation outcomes changes the
analysis before bootstrap choice even arises. Use a justified fitting/
selection split or cross-fitting design first. A conditional frozen-target
bootstrap does not include the entire fitting-and-selection procedure;
adaptive logging can require a time-aware argument beyond IID cluster resampling.

</details>
